# Кросс-валидация в `pytorch` для прогнозирования Seq2Seq

```python
import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset, random_split

import matplotlib.pyplot as plt

import pandas_datareader.data as web
```

```python
# Подготовка датасета для обучения в формате inputs, outputs
def make_datasets(input_data, n_inputs=2, n_outputs=1, gap=0):
	if input_data.ndim>1:
		input_data = input_data.squeeze()

	L = len(input_data)
	y = np.full((L-n_inputs-n_outputs-gap, n_outputs), 0.0)
	X = np.full((L-n_inputs-n_outputs-gap, n_inputs), 0.0)

	for i in range(n_inputs):
		X[:,i] = input_data[i:L-n_inputs-n_outputs-gap+i]

	for i in range(n_outputs):
		y[:,i] = input_data[n_inputs+gap+i:L-n_outputs+i]

	return X, y
```

```python
rate = web.DataReader(name='DGS10', data_source='fred', start='2000-01-01').dropna().diff().dropna()
rate.plot()
plt.show()
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device
```

```python
# Переведем ряда массив Numpy
series = rate.squeeze().to_numpy()
```

```python
# задаём ширину окна и горизонт прогнозирования
n_lags, fh= 30, 10

X, y = make_datasets(series, n_inputs=n_lags, n_outputs=fh)

X_tensor = torch.Tensor(X).unsqueeze(-1).to(device)
y_tensor = torch.Tensor(y).to(device)

dataset = TensorDataset(X_tensor, y_tensor)

# Для воспроизводимости задаём генератор
generator = torch.Generator().manual_seed(42)
# делим выборку на обучающую и тестовую с соотношении 80:20
train_dataset, test_dataset = random_split(dataset, [0.8, 0.2], generator=generator)

# Создаём DataLoader'ы
batch_size = 64

train_dataloader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_dataloader  = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
# Альтернативно
# Разделение на train/test
test_size = 0.8
test_len = int(test_size * X_tensor.size(dim=0))

X_train, X_test = X_tensor[:test_len], X_tensor[test_len+:]
y_train, y_test = y_tensor[:test_len], y_tensor[test_len:]

# Создание DataLoader
batch_size = 64

train_dataset = TensorDataset(X_train, y_train)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

test_dataset = TensorDataset(X_test, y_test)
test_loader = DataLoader(test_dataset, batch_size=batch_size)
```

```python
def train(train_loader, model, loss_fn, optimizer):
	train_loss = 0

	model.train()                                 # переводим модель в режим обучения
	for X_batch, y_batch in train_loader:
		X_batch, y_batch = X_batch.to(device), y_batch.to(device)

		optimizer.zero_grad()                     # Очищаем старые градиенты
		# Compute prediction error
		y_pred = model(X_batch)                         # строми прооноз на батче
		# print(f' X size: {X_batch.shape}, y_pred size: {y_pred.shape}, y_traget size: {y_batch.shape} \n')
		loss = loss_fn(y_pred, y_batch)                 # вычисляем ошибку проноза/значение функции потерь

		# Backpropagation
		loss.backward()                           # Вычисляем градиенты
		optimizer.step()                          # Обновляем веса по градиентам
		
		train_loss += loss.item()
	
	return train_loss / len(train_loader)
	# avg_train_loss = train_loss / len(train_loader)

def test(test_loader, model, loss_fn):
	model.eval()
	test_loss = 0

	with torch.no_grad():
		for X_batch, y_batch in test_loader:
			X_batch, y_batch = X_batch.to(device), y_batch.to(device)

			outputs = model(X_batch)
			loss = loss_fn(outputs, y_batch)
			test_loss += loss.item()

	return test_loss / len(test_loader)
	# avg_test_loss = test_loss / len(test_loader)
```

```python
hidden_state = 32

model = nn.Sequential(
	nn.RNN(input_size=1, hidden_size=hidden_state, num_layers=1, nonlinearity='tanh', bidirectional=False, batch_first=True),
	# nn.GRU(input_size=1, hidden_size=hidden_state, num_layers=1, bidirectional=False, batch_first=True),
	# nn.LSTM(input_size=1, hidden_size=hidden_state, num_layers=1, bidirectional=False, batch_first=True),
	ExtractHiddenState(),
	nn.Linear(hidden_state, fh)
).to(device)
```
